#### 2) Vamos montar um metadados

    Dica:
        1. inicie uma lista vazia

        3. conte o número de categorias dessa variável
        4. acumule essa informação de 3. na lista que você criou em 1.
        5. No final, essa lista pode ser atribuída à nossa variável.

````python
metadata = {coluna: dataset[coluna].dtype for coluna in dataset.columns}
````

````python
metadata
````

````python
metadata = pd.DataFrame({'nome_variavel': metadata.keys(), 'tipo_dado': metadata.values()})
categorias = []
for coluna in dataset.columns:
    if dataset[coluna].dtype == 'object':
        categorias.append(dataset[coluna].nunique())
    else:
        categorias.append(0)
metadata['qtd_categorias'] = categorias
display(metadata)
````

#### 3) Crie variáveis dummy para as variáveis necessárias (i.e. aquelas que são qualitativas e não estão armazenadas como {0, 1} ou {True, False}. Crie um *dataframe* apenas com as variáveis apropriadas para entrada no scikitlearn - elimine as variáveis tipo *str*, mantendo apenas suas versões *dummy*.

````python
variaveis_categoricas = [categorica for categorica in dataset.columns if dataset[categorica].dtype == 'object']
dataset_encoded = pd.get_dummies(dataset, columns=variaveis_categoricas, drop_first=True)
display(dataset_encoded)
````

````python
for coluna in dataset_encoded.columns:
  if dataset_encoded[coluna].dtype == 'bool':
    dataset_encoded[coluna] = dataset_encoded[coluna].astype(int)
````

````python
dataset_encoded.qt_pessoas_residencia = dataset_encoded.qt_pessoas_residencia.astype(int)
````

````python
dataset_encoded
````

#### 4) Qual variável é mais poderosa?

Considere as variáveis ```possui_email``` e ```posse_de_veiculo```. Faça uma tabela cruzada entre elas e responda qual delas te parece mais poderosa para prever a probabilidade de ```mau = 1```?

````python
# Cross-tabulation for 'possui_email' and 'mau'
crosstab_email = pd.crosstab(dataset_encoded['possui_email'], dataset_encoded['mau'], normalize='index')
print("Tabela cruzada: possui_email vs. mau")
display(crosstab_email)

# Probability of 'mau = 1' for 'possui_email'
prob_mau_email = crosstab_email[1].sort_values(ascending=False)
print("\nProbabilidade de mau = 1 por possui_email:")
display(prob_mau_email)

# Cross-tabulation for 'posse_de_veiculo_Y' (as 'posse_de_veiculo' is dummified) and 'mau'
crosstab_veiculo = pd.crosstab(dataset_encoded['posse_de_veiculo_Y'], dataset_encoded['mau'], normalize='index')
print("\nTabela cruzada: posse_de_veiculo_Y vs. mau")
display(crosstab_veiculo)

# Probability of 'mau = 1' for 'posse_de_veiculo_Y'
prob_mau_veiculo = crosstab_veiculo[1].sort_values(ascending=False)
print("\nProbabilidade de mau = 1 por posse_de_veiculo_Y:")
display(prob_mau_veiculo)
````

Ao analisar as tabelas de probabilidade de `mau = 1`:

*   **Possui Email:**
    *   Para quem *não possui email* (`possui_email = 0`), a probabilidade de `mau = 1` é de aproximadamente **2.38%**.
    *   Para quem *possui email* (`possui_email = 1`), a probabilidade de `mau = 1` é de aproximadamente **2.03%**.

*   **Possui Veículo:**
    *   Para quem *não possui veículo* (`posse_de_veiculo_Y = 0`, ou seja, 'N'), a probabilidade de `mau = 1` é de aproximadamente **2.43%**.
    *   Para quem *possui veículo* (`posse_de_veiculo_Y = 1`, ou seja, 'Y'), a probabilidade de `mau = 1` é de aproximadamente **2.21%**.

A variável `posse_de_veiculo` (ou `posse_de_veiculo_Y` após a dummyficação) parece ser um pouco mais poderosa para prever a probabilidade de `mau = 1` do que `possui_email`. Há uma diferença percentual maior na probabilidade de `mau = 1` entre os grupos que possuem e não possuem veículo (aproximadamente 0.67 pontos percentuais de diferença) em comparação com os grupos que possuem e não possuem email (aproximadamente 0.08 pontos percentuais de diferença). Isso sugere que a `posse_de_veiculo` tem uma separação mais distinta nos resultados de `mau = 1`.
